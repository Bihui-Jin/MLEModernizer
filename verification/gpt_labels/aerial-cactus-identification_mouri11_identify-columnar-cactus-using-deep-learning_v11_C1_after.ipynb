```python
import os
from os.path import join
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, Dropout, MaxPooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Fix: Use correct Kaggle input paths for this competition dataset
BASE_DIR = "../input/aerial-cactus-identification"
train_img_dir = join(BASE_DIR, "train")
test_img_dir = join(BASE_DIR, "test")
train_csv_path = join(BASE_DIR, "train.csv")
sample_sub_path = join(BASE_DIR, "sample_submission.csv")

# Basic sanity checks (fail fast with clear error if paths differ)
if not os.path.exists(train_img_dir):
    raise FileNotFoundError(f"Train image dir not found: {train_img_dir}")
if not os.path.exists(test_img_dir):
    raise FileNotFoundError(f"Test image dir not found: {test_img_dir}")
if not os.path.exists(train_csv_path):
    raise FileNotFoundError(f"Train CSV not found: {train_csv_path}")

train_data = pd.read_csv(train_csv_path)

# Use sample_submission order for test IDs to guarantee alignment
sample_sub = pd.read_csv(sample_sub_path)
test_ids = sample_sub["id"].tolist()
test_img_paths = [join(test_img_dir, img_id) for img_id in test_ids]

train_img_paths = [join(train_img_dir, img_id) for img_id in train_data["id"].tolist()]

img_size = 32


def prep_imgs(img_paths, img_height=img_size, img_width=img_size):
    # loading images and converting to numpy array
    imgs = [load_img(p, target_size=(img_height, img_width)) for p in img_paths]
    img_arr = np.array([img_to_array(img) for img in imgs], dtype=np.float32) / 255.0
    return img_arr


X_train = prep_imgs(train_img_paths)
y_train = train_data["has_cactus"].values  # sparse labels (0/1)

model = Sequential()
model.add(
    Conv2D(
        25,
        kernel_size=2,
        strides=2,
        activation="relu",
        input_shape=(img_size, img_size, 3),
    )
)
model.add(MaxPooling2D(pool_size=(2, 2), strides=2))
model.add(Conv2D(25, kernel_size=2, strides=2, activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2), strides=2))

model.add(Flatten())
model.add(Dense(250, activation="relu"))
model.add(Dense(2, activation="softmax"))

model.compile(
    loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"]
)
model.summary()

```

```python
train_datagen = ImageDataGenerator(
    featurewise_center=True,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
)
train_datagen.fit(X_train)

```

```python
# Fix: fit_generator is deprecated; model.fit works with generators in TF2
model.fit(
    train_datagen.flow(X_train, y_train, batch_size=64, shuffle=True),
    steps_per_epoch=len(X_train) // 64,
    epochs=3,
    verbose=2,
)
model.fit(X_train, y_train, batch_size=100, epochs=4, validation_split=0.2, verbose=2)

```

```python
X_test = prep_imgs(test_img_paths)

preds_temp = model.predict(X_test, batch_size=256, verbose=0)
# Fix: submission requires probability for "has_cactus" (class 1), not argmax class label
preds = preds_temp[:, 1].astype(np.float64)

output = pd.DataFrame({"id": test_ids, "has_cactus": preds})
output.to_csv("submission.csv", index=False)
print(output.head())
print("Wrote submission.csv with shape:", output.shape)
```
