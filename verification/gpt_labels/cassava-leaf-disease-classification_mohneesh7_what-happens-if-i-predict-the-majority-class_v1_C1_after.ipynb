```python
import os
import pandas as pd
from PIL import Image

```

```python
# Keep paths consistent with Kaggle dataset layout
work_dir = "../input/cassava-leaf-disease-classification/"
TRAIN_DIR = os.path.join(work_dir, "train_images")
TEST_DIR = os.path.join(work_dir, "test_images")

# Touch directory listing (as in original), but don't rely on it for submission ordering
_ = os.listdir(work_dir)
_ = os.listdir(TEST_DIR)

```

```python
data = pd.read_csv(os.path.join(work_dir, "train.csv"))

```

```python
data["label"].value_counts()

```

```python
# Change rationale: using the most frequent label from train is the smallest legitimate improvement
# over a fixed guess (still constant prediction, same core logic) and tends to increase accuracy.
majority_label = int(data["label"].value_counts().idxmax())

# Change rationale: generate submission rows in the exact order/content of sample_submission.csv
# to ensure a valid submission and stable scoring.
sample_sub = pd.read_csv(os.path.join(work_dir, "sample_submission.csv"))
test_image_ids = sample_sub["image_id"].tolist()

predictions = [majority_label] * len(test_image_ids)

sub = pd.DataFrame({"image_id": test_image_ids, "label": predictions})
sub.to_csv("submission.csv", index=False)

```
