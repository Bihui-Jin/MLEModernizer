```python
import pandas as pd
import numpy as np

```

```python
train_data = pd.read_csv("/kaggle/input/lmsys-chatbot-arena/train.csv")
test_data = pd.read_csv("/kaggle/input/lmsys-chatbot-arena/test.csv")

X_train = train_data[["id"]]
y_train = train_data[["winner_model_a", "winner_model_b", "winner_tie"]]
X_test = test_data[["id"]]

print(X_train.head())
print(y_train.head())
print(X_test.head())

```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

```

```python
# Logloss expects valid probabilities; LinearRegression can output negatives / >1.
# Minimal post-processing: clip to non-negative and renormalize to sum to 1 per row.
pred = np.clip(predictions, 0.0, None)
row_sums = pred.sum(axis=1, keepdims=True)

# If a row becomes all-zeros after clipping, fall back to uniform probabilities.
pred_proba = np.where(row_sums > 0, pred / row_sums, 1.0 / 3.0)

```

```python
# Fix submission column names to exactly match the required format (winner_tie, not winner_model_c).
submission = pd.DataFrame(
    {
        "id": test_data["id"].values,
        "winner_model_a": pred_proba[:, 0],
        "winner_model_b": pred_proba[:, 1],
        "winner_tie": pred_proba[:, 2],
    }
)

submission.head()

```

```python
submission.to_csv("submission.csv", index=False)
print("Wrote submission.csv with shape:", submission.shape)
print("Columns:", submission.columns.tolist())
```
