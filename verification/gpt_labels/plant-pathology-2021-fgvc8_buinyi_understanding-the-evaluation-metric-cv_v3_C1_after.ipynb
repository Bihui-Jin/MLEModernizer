```python
import numpy as np  # linear algebra
import pandas as pd  # data processing, CSV file I/O (e.g. pd.read_csv)
import os

```

```python
train = pd.read_csv("../input/plant-pathology-2021-fgvc8/train.csv")

# Change (score-toward-target): pick the RAREST label instead of the most frequent one.
# This keeps identical "constant-label baseline" semantics but should reduce mean F1
# (current score is above target, so we want to move downward).
label_counts = train["labels"].astype(str).str.split().explode().value_counts()

if len(label_counts):
    # Deterministic tie-break: among equally-rare labels, pick lexicographically first.
    min_count = label_counts.min()
    rare_labels = sorted(label_counts[label_counts == min_count].index.tolist())
    constant_label = rare_labels[0]
else:
    constant_label = "healthy"

print("Selected constant_label (rarest in train):", constant_label)

```

```python
TEST_FOLDER = "../input/plant-pathology-2021-fgvc8/test_images/"
images = os.listdir(TEST_FOLDER)
sub = pd.DataFrame(images, columns=["image"])

```

```python
sample_sub = pd.read_csv("../input/plant-pathology-2021-fgvc8/sample_submission.csv")
sub = sample_sub[["image"]].copy()

sub["labels"] = constant_label
sub.to_csv("submission.csv", index=False)
print("Wrote submission.csv with constant label:", constant_label, "| rows:", len(sub))
```
