```python
train = pd.read_csv('/kaggle/input/tabular-playground-series-may-2022/train.csv')
test = pd.read_csv('/kaggle/input/tabular-playground-series-may-2022/test.csv')
train.info()
```

```python
scaler = StandardScaler()
scaled_train = scaler.fit_transform(train.drop(['f_27', 'target'], axis=1))
```

```python
model = Sequential([
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='Adam', loss='binary_crossentropy', metrics=['AUC'])

callbacks = [EarlyStopping(patience=5, monitor='val_auc')]

history = model.fit(x=scaled_train, y=train['target'], validation_split=0.2, epochs=20, 
                    callbacks=callbacks, 
                    batch_size=64)
```

```python
scaled_test = scaler.transform(test.drop('f_27', axis=1))
```

```python
preds = model.predict(scaled_test)
test['target'] = (preds.reshape((len(test),)) > 0.5).astype('int')
```

```python
test.reset_index()[['id', 'target']].to_csv('submit.csv', index=False)
```
