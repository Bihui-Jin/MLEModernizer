```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import os
from collections import Counter
```

# Data Exploration

# Directories

```python
# Load the trained model and test directories
model_path_1 = '/kaggle/input/combinedmodel3/tensorflow2/default/1/BestModel_3454_8937.h5'
model_path_2 = '/kaggle/input/combinedmodel3/tensorflow2/default/1/best_model_0.37458707.h5'
model_path_3 = '/kaggle/input/combinedmodel3/tensorflow2/default/1/googlenet_inceptionv3.h5'
model_path_4 = '/kaggle/input/bestmodel_550_2/tensorflow2/default/1/BestModel_3577_8940.h5'
model_path_5 = '/kaggle/input/bestmodel_8878/tensorflow2/default/1/BestModel_8878_0358.h5'
test_image_dir = '/kaggle/input/cassava-leaf-disease-classification/test_images'
sample = '/kaggle/input/cassava-leaf-disease-classification/sample_submission.csv'
```

```python
sample_csv = pd.read_csv(sample)
sample_csv
```

# Load Model

```python
# Paths to each model and their respective input sizes
models_info = [
    (model_path_1, (550, 550)),
    (model_path_4, (550, 550)),
    (model_path_5, (512, 512))
]

# Load models with specified input sizes
models = [(load_model(path), input_size) for path, input_size in models_info]
```

```python
# Define class labels (optional, but useful for understanding the predictions)
class_labels = {
    0: 'Cassava Bacterial Blight (CBB)',
    1: 'Cassava Brown Streak Disease (CBSD)',
    2: 'Cassava Green Mottle (CGM)',
    3: 'Cassava Mosaic Disease (CMD)',
    4: 'Healthy'
}

# Prepare to store final predictions
image_predictions = []

# Loop through each image in the test directory
for image_id in os.listdir(test_image_dir):
    if image_id.endswith(('.jpg', '.jpeg', '.png')):  # Check for valid image formats
        # Initialize a list to collect predictions from each model
        model_predictions = []

        for model, input_size in models:
            # Load and preprocess the image for the model's input size
            image_path = os.path.join(test_image_dir, image_id)
            img = load_img(image_path, target_size=input_size)
            img_array = img_to_array(img) / 255.0
            img_array = np.expand_dims(img_array, axis=0)

            # Get the prediction from the model
            predictions = model.predict(img_array)
            predicted_class = np.argmax(predictions, axis=1)[0]
            model_predictions.append(predicted_class)

        # Perform majority voting
        final_predicted_class = Counter(model_predictions).most_common(1)[0][0]

        # Append the result
        image_predictions.append({
            'image_id': image_id,
            'label': final_predicted_class  # Kaggle expects numeric labels
        })

# Convert predictions to DataFrame for submission
submission_df = pd.DataFrame(image_predictions)

# Save the submission file
submission_path = '/kaggle/working/submission.csv'
submission_df.to_csv(submission_path, index=False)

print(f"Submission file saved at: {submission_path}")
```
