```python
import pandas as pd
import numpy as np

# Load the test set (final_df) and the training set (if needed later)
final_df = pd.read_csv("/kaggle/input/lmsys-chatbot-arena/test.csv")
train = pd.read_csv("/kaggle/input/lmsys-chatbot-arena/train.csv")
final_df.head()

```

```python
# Build the combined text column used in the original pipeline
final_df["text"] = (
    "User prompt: "
    + final_df["prompt"]
    + "\n\nModel A :\n"
    + final_df["response_a"]
    + "\n\n--------\n\nModel B:\n"
    + final_df["response_b"]
)
print(final_df["text"].iloc[0])

```

```python
# Instead of loading a pretrained GRU model (which is unavailable),
# we use a uniform probability baseline.
# For each of the three classes we assign probability 1/3.
num_samples = len(final_df)
uniform_prob = 1.0 / 3.0

final_df["winner_model_a"] = uniform_prob
final_df["winner_model_b"] = uniform_prob
final_df["winner_tie"] = uniform_prob

# Verify that each row sums to 1 (optional sanity check)
row_sums = final_df[["winner_model_a", "winner_model_b", "winner_tie"]].sum(axis=1)
assert np.allclose(row_sums, 1.0), "Row probabilities do not sum to 1."

final_df[["id", "winner_model_a", "winner_model_b", "winner_tie"]].head()

```

```python
# Write the submission file in the required format
final_df[["id", "winner_model_a", "winner_model_b", "winner_tie"]].to_csv(
    "submission.csv", index=False
)
print("Submission written to submission.csv")
```
