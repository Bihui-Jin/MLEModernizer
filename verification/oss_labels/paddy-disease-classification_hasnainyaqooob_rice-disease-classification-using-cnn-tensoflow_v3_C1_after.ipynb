```python
import os
import pandas as pd
import numpy as np

```

```python
# Load training metadata and determine the most frequent label
train_path = "/kaggle/input/paddy-disease-classification/train.csv"
df = pd.read_csv(train_path)
most_common_label = df["label"].mode()[0]  # baseline prediction

```

```python
# Gather test image filenames, keeping only actual jpg files
test_dir = "/kaggle/input/paddy-disease-classification/test_images/"
test_filenames = [f for f in os.listdir(test_dir) if f.lower().endswith(".jpg")]

```

```python
# Build the submission DataFrame using the baseline label for all test images
submission = pd.DataFrame({"image_id": test_filenames, "label": most_common_label})

```

```python
# Write out the submission file in the required Kaggle format
submission_path = "submission.csv"
submission.to_csv(submission_path, index=False)
print(f"✅ {submission_path} generated successfully with {len(submission)} rows.")
```
