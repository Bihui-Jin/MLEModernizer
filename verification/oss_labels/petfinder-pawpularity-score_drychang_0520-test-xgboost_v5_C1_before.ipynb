```python
import xgboost as xgb
import pandas as pd

# Load test data
df_test = pd.read_csv("/kaggle/input/petfinder-pawpularity-score/test.csv")

# Only keep metadata columns
metadata_features = [
    "Subject Focus", "Eyes", "Face", "Near", "Action",
    "Accessory", "Group", "Collage", "Human", "Occlusion",
    "Info", "Blur"
]
X_test = df_test[metadata_features].values

# Load your trained XGBoost model
model = xgb.XGBRegressor()
model.load_model("/kaggle/input/xgboost/other/default/1/xgb_model.json")

# Predict Pawpularity
preds = model.predict(X_test)

# Create submission file
submission = pd.DataFrame({
    "Id": df_test["Id"],
    "Pawpularity": preds
})
submission.to_csv("submission.csv", index=False)
```
