```python
import numpy as np
import pandas as pd
import os
import cv2
import random
from tqdm import tqdm
import seaborn as sns
from sklearn.preprocessing import LabelBinarizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import matplotlib.pyplot as plt


```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.callbacks import EarlyStopping

# Removed problematic import that raised AttributeError
from tensorflow.keras.utils import to_categorical


```

```python
train_dir = "../input/petfinder-pawpularity-score/train"
test_dir = "../input/petfinder-pawpularity-score/test"


```

```python
path0 = (
    "../input/petfinder-pawpularity-score/train/0007de18844b0dbbb5e1f607da0606e0.jpg"
)
image = cv2.imread(path0)
print(image.shape)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))


```

```python
image2 = cv2.resize(image, dsize=(60, 60), interpolation=cv2.INTER_CUBIC)
print(image2.shape)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))


```

```python
train = pd.read_csv("../input/petfinder-pawpularity-score/train.csv")
train


```

```python
train["Pawpularity"].unique()


```

```python
N0 = list(range(100))
N1 = list(range(1, 101))
normal_mapping = dict(zip(N1, N0))
reverse_mapping = dict(zip(N0, N1))


```

```python
train[train["Id"] == "0007de18844b0dbbb5e1f607da0606e0"]["Pawpularity"].tolist()[0]


```

```python
trainimg0 = []
trainlabel0 = []
for im in tqdm(os.listdir(train_dir)):
    # Build full path and read image
    img_path = os.path.join(train_dir, im)
    image = cv2.imread(img_path)
    # Skip if image could not be read
    if image is None:
        continue
    image2 = cv2.resize(image, dsize=(60, 60), interpolation=cv2.INTER_CUBIC)
    trainimg0.append(image2)
    # Extract label; skip if ID not found
    label_series = train.loc[train["Id"] == im[:-4], "Pawpularity"]
    if label_series.empty:
        continue
    trainlabel0.append(label_series.iloc[0])


```

```python
trainlabel1 = pd.Series(trainlabel0).map(normal_mapping)


```

```python
trainimg = np.array(trainimg0)
trainlabel = np.array(trainlabel1)


```

```python
m = len(trainimg)
M = list(range(m))
random.seed(2021)
random.shuffle(M)


```

```python
trainX = trainimg[M[0 : (m // 4) * 3]]
trainY0 = trainlabel[M[0 : (m // 4) * 3]]

testX = trainimg[M[(m // 4) * 3 :]]
testY0 = trainlabel[M[(m // 4) * 3 :]]


```

```python
labels1 = to_categorical(trainY0, num_classes=100)
trainY = np.array(labels1)


```

```python
trainx, testx, trainy, testy = train_test_split(
    trainX, trainY, test_size=0.2, random_state=44
)


```

```python
print(trainx.shape)
print(testx.shape)
print(trainy.shape)
print(testy.shape)


```

```python
model = Sequential()
model.add(Conv2D(32, (4, 4), input_shape=(60, 60, 3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Conv2D(64, (3, 3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.2))
model.add(Flatten())
model.add(Dense(400, activation="relu"))
model.add(Dense(100, activation="softmax"))


```

```python
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])


```

```python
model.summary()


```

```python
his = model.fit(
    trainx, trainy, validation_split=0.2, epochs=30, batch_size=92, verbose=2
)


```

```python
y_pred = model.predict(testx)
pred = np.argmax(y_pred, axis=1)
ground = np.argmax(testy, axis=1)
print(classification_report(ground, pred))


```

```python
get_acc = his.history["accuracy"]
value_acc = his.history["val_accuracy"]
get_loss = his.history["loss"]
validation_loss = his.history["val_loss"]

epochs = range(len(get_acc))
plt.plot(epochs, get_acc, "r", label="Accuracy of Training data")
plt.plot(epochs, value_acc, "b", label="Accuracy of Validation data")
plt.title("Training vs validation accuracy")
plt.legend(loc=0)
plt.figure()
plt.show()


```

```python
epochs = range(len(get_loss))
plt.plot(epochs, get_loss, "r", label="Loss of Training data")
plt.plot(epochs, validation_loss, "b", label="Loss of Validation data")
plt.title("Training vs validation loss")
plt.legend(loc=0)
plt.figure()
plt.show()


```

```python
pred2 = model.predict(testX)
print(pred2.shape)

PRED = []
for item in pred2:
    value2 = np.argmax(item)
    PRED += [value2]
print(pd.Series(PRED).value_counts())


```

```python
ANS = testY0
print(pd.Series(ANS).value_counts())
accuracy = accuracy_score(ANS, PRED)
print(accuracy)


```

```python
import seaborn as sns

fig, ax = plt.subplots(figsize=(14, 5))
sns.histplot(ANS, label="ANS", ax=ax, color="black", bins=100)
sns.histplot(PRED, label="PRED", ax=ax, color="C1", bins=100)
ax.legend()
ax.grid()
plt.show()


```

```python
fig, axs = plt.subplots(3, 3, figsize=(12, 12))
for i in range(9):
    r = i // 3
    c = i % 3
    img1 = testX[i]
    axs[r][c].axis("off")
    actual = reverse_mapping[testY0[i]]
    predict = reverse_mapping[PRED[i]]
    axs[r][c].set_title(str(actual) + "==" + str(predict))
    axs[r][c].imshow(img1)
plt.show()


```

```python
TESTX = []
testim = []
for im in tqdm(os.listdir(test_dir)):
    img_path = os.path.join(test_dir, im)
    image = cv2.imread(img_path)
    if image is None:
        continue
    image2 = cv2.resize(image, dsize=(60, 60), interpolation=cv2.INTER_CUBIC)
    TESTX.append(image2)
    testim.append(im[:-4])


```

```python
TESTX = np.array(TESTX)
print(TESTX.shape)


```

```python
test_pred2 = model.predict(TESTX)

TESTPRED = []
for item in test_pred2:
    value = np.argmax(item)
    value2 = reverse_mapping[value]
    TESTPRED.append(float(value2))
print(pd.Series(TESTPRED).value_counts())


```
