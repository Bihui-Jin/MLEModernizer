```python
import numpy as np  # linear algebra
import pandas as pd  # data processing, CSV file I/O (e.g. pd.read_csv)
import os
from collections import Counter
import pathlib

```

```python
train = pd.read_csv("../input/plant-pathology-2021-fgvc8/train.csv")
train.labels.value_counts()

```

```python
label_counter = Counter()
for lbls in train["labels"]:
    for lbl in lbls.split():
        label_counter[lbl] += 1
top_n = 1  # keep the baseline of predicting the single most common label
most_common_labels = [lbl for lbl, _ in label_counter.most_common(top_n)]
most_common_label_str = " ".join(most_common_labels)

```

```python
# Resolve the test image folder robustly (Kaggle notebooks may have a different cwd)
test_folder = pathlib.Path("../input/plant-pathology-2021-fgvc8/test_images")
if not test_folder.is_dir():
    test_folder = pathlib.Path("/kaggle/input/plant-pathology-2021-fgvc8/test_images")
images = [p.name for p in test_folder.iterdir() if p.is_file()]
sub = pd.DataFrame(images, columns=["image"])

```

```python
sub["labels"] = most_common_label_str
sub.to_csv("submission.csv", index=False)
```
