```python
import os
import shutil
import pydicom

root = '/kaggle/input/rsna-2022-cervical-spine-fracture-detection/test_images'
dcm_series = os.listdir(root)
rows=["row_id,fractured"]

for d in dcm_series:
    files = os.listdir(os.path.join(root,d))
    ds = pydicom.read_file(os.path.join(root,d,files[0]))
    sid = ds.StudyInstanceUID
    for k in ["C1","C2","C3","C4","C5","C6","C7"]:
        row = f"{sid}_{k},0.1"
        rows.append(row)
    for k in ["patient_overall"]:
        row = f"{sid}_{k},0.5"
        rows.append(row)
    
with open('submission.csv','w') as fp:
    fp.write('\n'.join(rows))
    
        
        
```
