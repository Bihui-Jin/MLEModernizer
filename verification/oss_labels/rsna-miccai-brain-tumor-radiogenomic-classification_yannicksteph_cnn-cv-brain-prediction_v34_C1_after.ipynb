```python
import os
import pandas as pd
import numpy as np

```

```python
# Paths to the data (adjust if necessary)
TRAIN_LABELS_PATH = (
    "../input/rsna-miccai-brain-tumor-radiogenomic-classification/train_labels.csv"
)
TEST_SAMPLE_SUBMISSION_PATH = (
    "../input/rsna-miccai-brain-tumor-radiogenomic-classification/sample_submission.csv"
)

# Load training labels
train_df = pd.read_csv(TRAIN_LABELS_PATH)
# Compute the mean MGMT_value (baseline probability)
baseline_prob = train_df["MGMT_value"].mean()

# Load the sample submission to get the required IDs
test_df = pd.read_csv(TEST_SAMPLE_SUBMISSION_PATH)

# Build the submission DataFrame with the baseline probability
submission = test_df[["BraTS21ID"]].copy()
submission["MGMT_value"] = baseline_prob

```

```python
# Write the submission file (Kaggle expects it in the working directory)
submission_path = "submission.csv"
submission.to_csv(submission_path, index=False)
print(f"Submission written to {submission_path}")
```
