```python
import numpy as np  # linear algebra
import pandas as pd  # data processing, CSV file I/O (e.g. pd.read_csv)
from sklearn.model_selection import train_test_split
import os

for dirname, _, filenames in os.walk("/kaggle/input"):
    for filename in filenames:
        print(os.path.join(dirname, filename))

```

```python
df_train = pd.read_csv("../input/tabular-playground-series-dec-2021/train.csv")
df_test = pd.read_csv("../input/tabular-playground-series-dec-2021/test.csv")


```

```python
def reduce_mem_usage(df, verbose=True):
    numerics = ["int16", "int32", "int64", "float16", "float32", "float64"]
    start_mem = df.memory_usage().sum() / 1024**2
    for col in df.columns:
        col_type = df[col].dtypes
        if col_type in numerics:
            c_min = df[col].min()
            c_max = df[col].max()
            if str(col_type)[:3] == "int":
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                else:
                    df[col] = df[col].astype(np.int64)
            else:
                if (
                    c_min > np.finfo(np.float32).min
                    and c_max < np.finfo(np.float32).max
                ):
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float64)
    end_mem = df.memory_usage().sum() / 1024**2
    if verbose:
        print(
            "Mem. usage decreased to {:5.2f} Mb ({:.1f}% reduction)".format(
                end_mem, 100 * (start_mem - end_mem) / start_mem
            )
        )
    return df


```

```python
df_train = reduce_mem_usage(df_train)
df_test = reduce_mem_usage(df_test)

```

```python
def is_categorical(data, column):
    if len(data[column].unique()) <= 15:
        print(str(column) + ": " + str(data[column].unique()))
    return None


```

```python
for column in df_train.columns:
    is_categorical(df_train, column)

```

```python
df_train = df_train.drop(["Soil_Type7", "Soil_Type15"], axis=1)
df_test = df_test.drop(["Soil_Type7", "Soil_Type15"], axis=1)

```

```python
targets = df_train["Cover_Type"] - 1
df_train = df_train.drop(["Cover_Type"], axis=1)

```

```python
targets.value_counts()

```

```python
from sklearn import preprocessing

scaler = preprocessing.MinMaxScaler()

numeric_features = df_train.columns[1:11].tolist()

df_train[numeric_features] = scaler.fit_transform(df_train[numeric_features])
df_test[numeric_features] = scaler.transform(df_test[numeric_features])

```

```python
# Fixed split: removed stratify (which caused a single‑sample class error) and set an explicit test size.
train_X, val_X, train_y, val_y = train_test_split(
    df_train, targets, test_size=0.2, random_state=1, shuffle=True
)

```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

model = XGBClassifier(
    tree_method="hist",
    learning_rate=0.3,
    use_label_encoder=False,
    eval_metric="mlogloss",
)

model.fit(train_X, train_y, verbose=False)

val_pred = model.predict(val_X)
acc = accuracy_score(val_y, val_pred)
print(f"accuracy_score: {acc}")

```

```python
submission = pd.read_csv(
    "../input/tabular-playground-series-dec-2021/sample_submission.csv"
)

test_pred = model.predict(df_test) + 1  # shift back to original 1‑7 label range
submission["Cover_Type"] = test_pred.astype(int)

submission.to_csv("submission.csv", index=False)
```
