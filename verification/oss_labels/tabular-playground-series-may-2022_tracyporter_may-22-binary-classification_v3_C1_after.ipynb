```python
target = train["target"]
target

```

```python
# combine train (without target) and test; use concat instead of deprecated append
combi = pd.concat([train.drop(columns=["target"]), test], axis=0, ignore_index=True)
# drop id and a possibly‑non‑existent feature safely
combi = combi.drop(columns=["id", "f_27"], errors="ignore")
combi

```

```python
corr = combi.corr()
f, ax = plt.subplots(figsize=(12, 9))
sns.heatmap(corr, vmax=0.8, square=True)

```

```python
print(corr)

```

```python
columns = np.full((corr.shape[0],), True, dtype=bool)
for i in range(corr.shape[0]):
    for j in range(i + 1, corr.shape[0]):
        if corr.iloc[i, j] >= 0.80:
            if columns[j]:
                columns[j] = False
selected_columns = combi.columns[columns]
combi = combi[selected_columns]
combi

```

```python
combi = (combi - combi.min()) / (combi.max() - combi.min())
combi

```

```python
y = target
X = combi[: len(train)]
X_test = combi[len(train) :]

```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.1, random_state=42)
X_train.shape, X_val.shape, y_train.shape, y_val.shape, X_test.shape

```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42, max_iter=1000, n_jobs=-1).fit(
    X_train, y_train
)
print(model.score(X_train, y_train))

```

```python
y_pred = model.predict(X_val)
print(model.score(X_val, y_val))

```

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_val, y_pred))

```

```python
# Use predicted probabilities for the positive class (required for AUC / submission)
preds = model.predict_proba(X_test)[:, 1]
preds

```

```python
submission["target"] = preds
submission.to_csv("/kaggle/working/submission.csv", index=False)
submission = pd.read_csv("/kaggle/working/submission.csv")
submission
```
